```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
from imblearn.over_sampling import SMOTE
from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_class_weight
import numpy as np
```

```python
# 1. Cargar dataset
ruta = r"C:\Users\USER\OneDrive - universidadean.edu.co\2. Semestre\Seminario de investigación\Modulo 2\Modelo ML\data\processed\df_preprocesado_XGBoost.csv"
df = pd.read_csv(ruta)
```

```python
# 2. Separar variables predictoras y objetivo
X = df.drop(columns=['MORA_ACTUAL', 'MORA_SIGUIENTE', 'ETIQUETA_MORA'])
y = df['ETIQUETA_MORA'].astype(int)
```

```python
# Visualización de distribución de clases
plt.figure(figsize=(6, 4))
sns.countplot(x=y)
plt.title("Distribución de clases - ETIQUETA_MORA")
plt.xlabel("Clase")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# División en train y test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
# Aplicar SMOTE solo sobre el train
smote = SMOTE(random_state=42)
X_train_bal, y_train_bal = smote.fit_resample(X_train, y_train)
```

```python
# Visualización de distribución de clases DESPUÉS del balanceo
plt.figure(figsize=(6, 4))
sns.countplot(x=y_train_bal)
plt.title("Distribución de clases - DESPUÉS de SMOTE")
plt.xlabel("Clase")
plt.ylabel("Frecuencia")
plt.show()
```

```python
excluir_manual = [
    'NUMERO DE LA COLOCACION', 'NUMERO DE IDENTIFICACION DEUDOR',
    'NUMERO DEL PRODUCTO', 'NUMERO DE LA AGENCIA',
    'MORA_ACTUAL', 'MORA_SIGUIENTE', 'ETIQUETA_MORA'
]

# Agregar columnas binarias (dummies) a excluir del escalado
excluir_dummies = [col for col in X_train.columns if (
    X_train[col].nunique() == 2 and set(X_train[col].unique()) <= {0, 1}
)]

# Columnas a escalar (numéricas reales, no dummies ni ID)
variables_a_escalar = [
    col for col in X_train.select_dtypes(include=['int64', 'float64']).columns
    if col not in excluir_manual + excluir_dummies
]

# Escalamiento
scaler = StandardScaler()

# Convertir arrays a DataFrame para conservar nombres de columnas
X_train_bal_df = pd.DataFrame(X_train_bal, columns=X_train.columns)
X_test_df = X_test.copy()

# Aplicar escalamiento solo a columnas seleccionadas
X_train_bal_df[variables_a_escalar] = scaler.fit_transform(X_train_bal_df[variables_a_escalar])
X_test_df[variables_a_escalar] = scaler.transform(X_test_df[variables_a_escalar])
```

```python
# Definir hiperparámetros
xgb_model = XGBClassifier(
    objective='multi:softmax',
    num_class=4,
    eval_metric='mlogloss',
    use_label_encoder=False,
    random_state=42,
    n_estimators=300,
    max_depth=8,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=1.0,
    min_child_weight=3
)

# Ajustar pesos de clase
clases = np.array([0, 1, 2, 3])
pesos = compute_class_weight(class_weight='balanced', classes=clases, y=y_train_bal)
class_weights = dict(zip(clases, pesos))
sample_weights = [class_weights[y] for y in y_train_bal]

# Entrenamiento con sample_weight
xgb_model.fit(X_train_bal_df, y_train_bal, sample_weight=sample_weights)

# Predicción y evaluación final
y_pred = xgb_model.predict(X_test_df)

print("\neporte de Clasificación (modelo con hiperparámetros manuales):")
print(classification_report(y_test, y_pred, target_names=[
    "0: Al día", "1: Entra mora", "2: Sale mora", "3: Se mantiene mora"
]))

# Matriz de Confusión
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=["0: Al día", "1: Entra mora", "2: Sale mora", "3: Se mantiene mora"],
            yticklabels=["0: Al día", "1: Entra mora", "2: Sale mora", "3: Se mantiene mora"])
plt.title("Matriz de Confusión - XGBoost (Hiperparámetros manuales)")
plt.xlabel("Predicho")
plt.ylabel("Real")
plt.tight_layout()
plt.show()
```
